### Handle Missing Values

```python
df["Study_Hours"]=df["Study_Hours"].fillna(df["Study_Hours"].mean())
df["Attendance"]=df["Attendance"].fillna(df["Attendance"].mean())
df["Science"]=df["Science"].fillna(df["Science"].mean())
```

### Gender Graph Distribution (Bar Chart)

```python
plt.figure(figsize=(7,5))

gender_count = df['Gender'].value_counts()

plt.bar(gender_count.index, gender_count.values,
        color=['#9B59B6', '#3498DB'],
        edgecolor='black', width=0.5)

plt.title('Gender Distribution', fontsize=16)
plt.xlabel('Gender')
plt.ylabel('Number of Students')

plt.bar_label(plt.gca().containers[0], fontsize=12)

plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Study Hours Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(df['Study_Hours'], bins=10,
         color='#831C91',
         edgecolor='black',
         alpha=0.85)

plt.title('Study Hours Distribution', fontsize=16)
plt.xlabel('Study Hours')
plt.ylabel('Number of Students')

plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Attendace Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(df['Attendance'], bins=10, color='skyblue',
         edgecolor='black', alpha=0.85)

plt.title('Attendance Distribution', fontsize=16)
plt.xlabel('Attendance (%)')
plt.ylabel('Number of Students')

plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Marks Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(df['Math'], bins=10,
         color='#F39C12',
         edgecolor='black',
         alpha=0.85)

plt.title('Math Marks Distribution', fontsize=16)
plt.xlabel('Math Marks')
plt.ylabel('Number of Students')

plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Science'], bins=10,
         color='#FF69B4',
         edgecolor='black',
         alpha=0.85)

plt.title('Science Marks Distribution', fontsize=16)
plt.xlabel('Science Marks')
plt.ylabel('Number of Students')

plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['English'], bins=10,
         color='#20B2AA',
         edgecolor='black',
         alpha=0.85)

plt.title('English Marks Distribution', fontsize=16)
plt.xlabel('English Marks')
plt.ylabel('Number of Students')

plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))

subjects = ['Math', 'Science', 'English']

average_marks = [
    df['Math'].mean(),
    df['Science'].mean(),
    df['English'].mean()
]

bars = plt.bar(
    subjects,
    average_marks,
    color=['#C3B1E1', '#FFB38A', '#A8E6CF'],
    edgecolor='black',
    width=0.5
)

plt.title('Subject-wise Average Marks', fontsize=16)
plt.xlabel('Subjects')
plt.ylabel('Average Marks')

plt.bar_label(bars, fmt='%.2f')
plt.grid(axis='y', alpha=0.3)

plt.show()
```

### Study Hours vs Math Marks

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Study_Hours'], df['Math'],
            color='#8E44AD',
            edgecolor='black',
            alpha=0.7,
            s=60)

plt.title('Study Hours vs Math Marks', fontsize=16)
plt.xlabel('Study Hours')
plt.ylabel('Math Marks')

plt.grid(alpha=0.3)
plt.show()
```

### Attandance vs Average Marks

```python
df['Average_Marks'] = df[['Math', 'Science', 'English']].mean(axis=1)

plt.figure(figsize=(8,5))

plt.scatter(df['Attendance'], df['Average_Marks'],
            color='#E67E22',
            edgecolor='black',
            alpha=0.7,
            s=60)

plt.title('Attendance vs Average Marks', fontsize=16)
plt.xlabel('Attendance (%)')
plt.ylabel('Average Marks')

plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,6))

corr = df[['Study_Hours', 'Attendance', 'Math',
           'Science', 'English']].corr()

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm',
            linewidths=0.5,
            fmt='.2f')

plt.title('Student Performance Correlation Heatmap', fontsize=14)
plt.show()
```

```python
print("Total Students:", len(df))

print("Gender Distribution:")
print(df['Gender'].value_counts())

print("\nAverage Study Hours:")
print(df['Study_Hours'].mean())

print("\nAverage Attendance:")
print(df['Attendance'].mean())

print("\nAverage Subject Marks:")
print(df[['Math', 'Science', 'English']].mean())
```

```python
# Correlation Analysis

correlation = df[['Study_Hours', 'Attendance', 'Math', 'Science', 'English']].corr()

print("Correlation with Math Marks:")
print(correlation['Math'].sort_values(ascending=False).round(2))

print("\nCorrelation with Science Marks:")
print(correlation['Science'].sort_values(ascending=False).round(2))

print("\nCorrelation with English Marks:")
print(correlation['English'].sort_values(ascending=False).round(2))
```

### Key Findings
Key Findings

- The dataset contains 150 students, including 87 male and 63 female students.
- The average study time of students is approximately 4.39 hours.
- The average attendance of students is 77.25%.
- The average Mathematics score is 57.80.
- Study hours show a positive correlation with Mathematics (0.59), Science (0.61), and English (0.70) scores.
- English has the strongest relationship with study hours among the three subjects.
- Attendance shows a weak positive correlation with subject marks.
- Data visualization helped identify patterns and relationships in student academic performance.

Limitations
